```python
import earthkit.data as ekd
import earthkit.plots as ekp
import os
os.environ["ECCODES_VERSION_CHECK_OFF"] = "1"


path = "/store_new/mch/msopr/osm/ICON-CH1-EPS/FCST25/25100500_638/grib/i1eff00030000_000"

# Load the local GRIB file
data = ekd.from_source("file", path).to_fieldlist()
```

```python
CLCT = data.sel({"parameter.variable": "CLCT"})
CLCT.ls()
```

```python
chart = ekp.Map(domain="Switzerland")
chart.contourf(CLCT)
chart.coastlines()
chart.borders()
chart.legend()
chart.title("ICON-CH1-EPS total cloud cover")
chart.show()
```
